```python
sns.get_dataset_names()
```

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins.dropna(inplace=True)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(penguins['bill_length_mm'], penguins['body_mass_g'])
```

```python
plt.scatter(penguins['bill_depth_mm'], penguins['body_mass_g'])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
data = penguins[['bill_length_mm', 'bill_depth_mm']]
target = penguins['body_mass_g']
```

```python
type(data)
```

```python
type(target)
```

```python
data_train, data_test, target_train, target_test = train_test_split(data, target, test_size=0.2)
```

```python
lr_model = LinearRegression()
```

```python
lr_model.fit(data_train, target_train)
```

```python
lr_model.coef_
```

```python
lr_model.intercept_
```

```python
y_pred = lr_model.predict(data_test)
```

```python
y_pred[:5]
```

```python
target_test[:5]
```

```python
from sklearn.metrics import root_mean_squared_error

rmse = root_mean_squared_error(target_test, y_pred)
```

```python
rmse
```
